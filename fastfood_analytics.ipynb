```python
restaurant_calories = df.groupby("Restaurant")["Calories"].mean()

plt.figure(figsize=(8,5))

# Create a list of colors in the order of restaurant_calories.index
bar_colors = []
for r_name in restaurant_calories.index:
    # Adjust names to match dictionary keys if necessary
    if r_name == 'McDonalds':
        dict_key = "McDonald's"
    elif r_name == 'Dominos':
        dict_key = "Domino's"
    else:
        dict_key = r_name
    bar_colors.append(color.get(dict_key, '#AAAAAA')) # Use a default grey for any unmatched names

plt.bar(
    restaurant_calories.index,
    restaurant_calories.values,
    color=bar_colors
)

plt.title("Average Calories by Restaurant")

plt.xlabel("Restaurant")

plt.ylabel("Average Calories")

plt.xticks(rotation=30)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
top_protein = df.sort_values("Protein", ascending=False).head(10)

plt.figure(figsize=(8,5))

plt.barh(
    top_protein["Item"],
    top_protein["Protein"]
)

plt.title("Top Protein Rich Foods")

plt.xlabel("Protein (g)")
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
